# About the Study Habits Dataset

## Introduction

This project explores a **Study Habits Survey Dataset** containing information about 80 students. The purpose is to understand students' study habits and investigate how factors such as study time, sleeping hours, phone usage, and daily travel relate to their academic performance (GPA).

## Dataset Information

The dataset contains the following information:

- **Number of students:** 80
- **Number of columns:** 9
- **Data source:** Simulated student survey data.
- **Analysis tool:** Python using the pandas library.
- **Notebook environment:** Jupyter Notebook.

## Description of the Dataset Columns

| Column Name | Data Type / Unit | Description |
|---|---|---|
| student_id | Text | Unique identification number for each student. |
| faculty | Category | Faculty where the student studies. |
| year | Integer | Student's year of study (1–4). |
| age | Years | Age of the student. |
| study_hours_per_week | Hours | Number of hours spent studying per week. |
| sleep_hours | Hours | Average sleeping hours per night. |
| commute_minutes | Minutes | Time spent travelling to campus each day. |
| phone_hours_per_day | Hours | Number of hours spent using a phone each day. |
| gpa | Number (0–4) | Grade Point Average representing academic performance. |

## Objective of the Analysis

The main objective is to explore the dataset and identify possible relationships between students' study habits and academic performance. The analysis will help us understand whether study hours, sleep, phone usage, and other factors are associated with students' GPA.

## 4. Asking questions*Task 4: follow the pattern Question, Code, Output, Interpretation.***Q1. What is the average GPA of all students?**

```python
df["gpa"].mean()
```

## The average GPA of all 80 students is 2.581, which is approximately 2.58 out of 4.

```python
# Q2. Average study hours per faculty
df.groupby("faculty")["study_hours_per_week"].mean()
```

## Engineering students study the most, averaging 14.93 hours per week, while Education students study the least, averaging 11.97 hours per week.

```python
# Q3. How many students use the phone for more than 5 hours a day?
df[df["phone_hours_per_day"] > 5].shape[0]
```

## number of student how use phone at least five hour are 4

```python
# Q4. Are study hours and GPA related? (correlation: -1 to +1)
df[["study_hours_per_week", "gpa"]].corr()
```

## There is a positive correlation of 0.57 between study hours and GPA, meaning students who study more tend to have higher GPAs.

```python
# Bonus: the five highest GPAs
df.sort_values("gpa", ascending=False).head(5)
```

## The five students have the highest GPAs, with student S046 having the highest GPA of 3.39 out of 4.

## 5. Charts*Task 5: make two simple charts and add a one-sentence caption under each.*

```python
import matplotlib.pyplot as plt

# GPA Histogram
df["gpa"].plot.hist(bins=5, title="Distribution of GPA")
plt.show()
```

## The chart shows the distribution of students' GPAs in 5 groups

```python
df.plot.scatter(x="study_hours_per_week", y="gpa",
                title="Study hours vs GPA")
plt.show()
```
